```python
from sklearn.naive_bayes import GaussianNB
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split


iris = load_iris()
X, y = iris.data, iris.target


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

gnb = GaussianNB()


gnb.fit(X_train, y_train)

predictions = gnb.predict(X_test)


accuracy = gnb.score(X_test, y_test)
print(f"Accuracy (in-built function): {accuracy}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV

iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


knn = KNeighborsClassifier()

k_range = list(range(1, 31))  # Example: check k values from 1 to 30
param_grid = dict(n_neighbors=k_range)

# 5. Create a GridSearchCV object
grid = GridSearchCV(knn, param_grid, cv=5, scoring='accuracy')  

# 6. Fit the GridSearchCV object to the training data
grid.fit(X_train, y_train)

# 7. Print the best hyperparameter value and the corresponding accuracy
print("Best k:", grid.best_params_['n_neighbors'])
print("Best cross-validation score:", grid.best_score_)

# 8. Evaluate the model with the best hyperparameter on the test set
accuracy = grid.score(X_test, y_test)
print("Test set accuracy:", accuracy)
```
